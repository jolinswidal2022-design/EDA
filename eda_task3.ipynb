```python
df=df.drop_duplicates()
```

```python
df["Price_Delta"] = df["Close"] - df["Open"]
df
```

```python
df["Daily_Return"]=((df["Close"]-df["Open"]))/df["Open"]*100
df
```

```python
df["Volume"].mean()
```

```python
df["Volume"].max()
```

```python
df["Volume_MA_7"]=df["Volume"].rolling(7).mean() #moving average of volume
df
```

```python
Volume_mean=df["Volume"].mean()#identifying anomolies
```

```python
Volume_mean
```

```python
Volume_std=df["Volume"].std()
```

```python
Volume_std
```

```python
Volume_threshold=Volume_mean+(2*Volume_std)
Volume_threshold
```

```python
anomalies=df[df["Volume"]>Volume_threshold]
anomalies
    
```

```python
#analyse stock return distribution
mean_return=df["Daily_Return"].mean()
print("Mean Return:",mean_return)

variance_return=df["Daily_Return"].var()
print("var return:",variance_return)

std_return=df["Daily_Return"].std()
print("std return:",std_return)
```
