```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"])
df.dtypes
```

```python
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"])
df.dtypes
```

```python
#scaterplot 
df_pivot=pd.pivot_table(df,
    index="Medical Condition",
    columns="Insurance Provider",
    values="Billing Amount",
    aggfunc="sum"
)
print(df_pivot)
```

```python
df_pivot.plot(kind="bar",stacked=True,figsize=(12,5))
plt.title("plots comparing billing amounts across medical conditions and insurance providers")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
#violin plots
sns.violinplot(x="Insurance Provider",y="Billing Amount",data=df)
plt.title("Billing amount by insurance provider")
plt.ylabel("Billing Amount")
plt.xlabel("Insurance Provider")
plt.xticks(rotation=55)
plt.show()
```

```python
df["Hospital stay days"]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
df["Hospital stay days"]
```

```python
numeric_column=["Age","Hospital stay days","Billing Amount"]
```

```python
correlations=df[numeric_column].corr()
```

```python
correlations
```

```python
sns.heatmap(correlations,
            annot=True,
            cmap="coolwarm",
            cbar=False,
            square=True,
            linewidth=2)
plt.title("Correlation Matrix of Age, Stay Duration and Billing Amount")
plt.show()
```
