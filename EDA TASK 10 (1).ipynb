```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("kc_house_data.csv")
```

```python
df
```

```python
plt.figure(figsize=(10,6))
sns.regplot(data=df,x="sqft_living",y="price")
plt.title("sqft living vs price")
plt.xlabel("sqft living")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x="waterfront",y="price")
plt.title("waterfront vs price")
plt.xlabel("waterfront")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="lat",
    y="long"
)
plt.title("lattitude vs longitude")
plt.xlabel("lattitude")
plt.ylabel("longitude")
plt.legend=False
plt.palette="red"
plt.show()
```

```python
locate_summary=df.groupby("zipcode")["price"].agg(
    ["count", "mean", "median", "max"]
).reset_index()
locate_summary=locate_summary.sort_values(
by="mean",
ascending=False
)
locate_summary.head(10)
```
